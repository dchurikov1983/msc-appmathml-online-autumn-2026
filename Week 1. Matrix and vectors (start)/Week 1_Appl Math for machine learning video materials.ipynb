# О списках в Python

Списки (`list`) — основной инструмент.
Списки могут хранить элементы любых типов (числа, строки, другие списки) и динамически изменяют размер.

Создание:

```python
my_list = [1, 2, 3, "hello", True]  # Список с разными типами данных
matrix = [[1, 2], [3, 4]]           # Двумерный список (матрица)
```

Обращение к элементам:

```python
print(my_list[0])   # 1 (индексация с нуля)
print(matrix[1][0]) # 3 (доступ к элементу вложенного списка)
print(my_list[-1])  # True (отрицательный индекс — отсчёт с конца)
```

Изменение элементов:

```python
my_list[3] = "world"  # Замена элемента → [1, 2, 3, "world", True]
my_list.append(4)     # Добавление в конец → [1, 2, 3, "world", True, 4]
my_list
```

Срезы:

```python
print(my_list[1:4])  # [2, 3, "world"] (элементы с индексом 1, 2, 3)
```

#Работа с матрицами в библиотеке NumPy

NumPy — библиотека для работы с многомерными массивами данных для Python.

- Вектора — одномерные массивы.
- Матрицы — двумерные массивы.
- Тензоры — $\dots$

Функционал библиотеки сильно упрощает работу с массивами по сравнению с использованием списков. Большинство библиотек, в которых предполагается работа с многомерными данными, поддерживают NumPy-подобный синтаксис.

Благодаря транслированию Python-кода в С-код, выполнение операций в NumPy происходит значительно быстрее, чем при использовании обычных Python-списков. Некоторые функции также поддерживают распараллеливание на несколько ядер.

### Индексация

В NumPy такая же индексация и логика срезов, как и в целом в Python. **Индексация также начинается с 0.**

#### Одномерное индексирование

```python
x = np.arange(1, 10, 1)
x
```

```python
x[3] # взять конкретный элемент
```

```python
y = x[:] # взять все
```

```python
x[2:-1:2] #  взять со 2-го по −1-й с шагом 2
```

```python
x[::-1] # перевернуть массив
```

В NumPy, в отличие от Python, можно брать определённые индексы, передавая их через список:

```python
x[[0, 7, 5]]
```

Аналогичный код на Python:

```python
y = list(range(1, 10, 1))
y[[0, 7, 5]]
```

#### Многомерное индексирование

Так как NumPy работает с многомерными матрицами, то и многомерное индексирование и срезы тут тоже поддерживаются. Выглядит это всё по аналогии с одномерным индексированием.  

```python
X = x.reshape((3, 3))
X
```

Каждое измерение отделяется запятой, и в каждом измерении можно проводить независимые индексации:

```python
X[0, 2] # взять конкретный элемент по конкретному набору индексов
```

```python
X[:, 2] # взять все строки (первое измерение), взять только второй столбец (второе измерение)
```

```python
X[:, 2].shape
```

```python
X[:, 2:3] # а так сохранится размерность
```

```python
X[:, 2:3].shape
```

Срезы можно использовать и в нескольких измерениях:

```python
X[:2, :2] # можно взять верхнюю левую подматрицу размера 2 × 2
```

Также можно передавать координаты нужных элементов в виде списков координат по каждой из осей.

```python
X, type(X)
```

```python
X[[0, 0, 2], [0, 2, 1]]
```

При такой записи возьмутся элементы на позициях (0, 0), (0, 2), (2, 1) — нужно читать два списка паралелльно и создавать из них координаты элемента.
